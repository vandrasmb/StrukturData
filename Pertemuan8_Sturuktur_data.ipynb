```python
#Implementasi BFS
from collections import deque

def bfs(graph, start):
    visited = set()
    queue = deque([start])
    result = []

    while queue:
        node = queue.popleft()
        if node not in visited:
            visited.add(node)
            result.append(node)
            for neighbor in graph[node]:
                if neighbor not in visited:
                    queue.append(neighbor)
    return result
```

```python
#Implementasi DFS
def dfs(graph, start):
    visited = set()
    result = []

    def explore(node):
        visited.add(node)
        result.append(node)
        for neighbor in graph[node]:
            if neighbor not in visited:
                explore(neighbor)

    explore(start)
    return result
```

```python
#Fungsi find_path(graph, start, end) (DFS)
def find_path(graph, start, end):
    visited = set()
    path = []

    def dfs_path(node):
        visited.add(node)
        path.append(node)

        if node == end:
            return True

        for neighbor in graph[node]:
            if neighbor not in visited:
                if dfs_path(neighbor):
                    return True

        path.pop()
        return False

    if dfs_path(start):
        return path
    else:
        return None
```

```python
#Fungsi is_connected(graph)
def is_connected(graph):
    nodes = list(graph.keys())
    if not nodes:
        return True

    visited = set()

    def dfs(node):
        visited.add(node)
        for neighbor in graph[node]:
            if neighbor not in visited:
                dfs(neighbor)

    dfs(nodes[0])

    return len(visited) == len(nodes)
```
