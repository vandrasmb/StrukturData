```python
class Node:
  def __init__(self, data):
    self.data = data
    self.left = None
    self.right =  None

root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right =  Node(5)

def inorder(node):
  if node:
    inorder(node.left)
    print(node.data, end=" ")
    inorder(node.right)

print("Hasil inorder traversal dari binary tree:")
inorder(root)
```

```python
class BinarySearchTree:
  def __init__(self):
    self.root = None

  def insert(self, root, data):
    if root is None:
      return Node(data)
    if data < root.data:
      root.left = self.insert(root.left, data)
    else:
      root.right = self.insert(root.right, data)
    return root

  # Add an inorder traversal method to the BST class
  def inorder_traversal(self, root):
    if root:
      self.inorder_traversal(root.left)
      print(root.data, end=" ")
      self.inorder_traversal(root.right)

# Re-initialize and build the BST with the new inorder_traversal method
bst = BinarySearchTree()
root = None
data_list = [50, 30, 70, 20, 40, 60, 80]
for value in data_list:
  root = bst.insert(root, value)

print("Inorder traversal of the BST:")
bst.inorder_traversal(root)
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

class BinarySearchTree:
    def __init__(self):
        self.root = None

    # Fungsi menyisipkan data sesuai aturan BST
    def insert(self, root, data):
        if root is None:
            return Node(data)
        if data < root.data:
            root.left = self.insert(root.left, data)
        else:
            root.right = self.insert(root.right, data)
        return root

    # Fungsi pencarian nilai
    def search(self, node, key):
        if node is None or node.data == key:
            return node
        if key < node.data:
            return self.search(node.left, key)
        return self.search(node.right, key)

# --- Implementasi Tugas ---
bst = BinarySearchTree()
root_bst = None

# Daftar 7 angka unik (NPM, TTL, ASCII Nama, Angka Bebas)
data_list = [26, 18, 10, 114, 97, 10, 88]

for value in data_list:
    root_bst = bst.insert(root_bst, value)

# Menampilkan hasil traversal
def print_traversal(node, type="inorder"):
    if node:
        if type == "preorder": print(node.data, end=' ')
        print_traversal(node.left, type)
        if type == "inorder": print(node.data, end=' ')
        print_traversal(node.right, type)
        if type == "postorder": print(node.data, end=' ')

print("\nHASIL TRAVERSAL BST")
print("Inorder   :", end=' '); print_traversal(root_bst, "inorder")
print("\nPreorder  :", end=' '); print_traversal(root_bst, "preorder")
print("\nPostorder :", end=' '); print_traversal(root_bst, "postorder")

# Fitur Pencarian
print("\n\nHASIL PENCARIAN")
key_npm = 26
result = bst.search(root_bst, key_npm)
print(f"Mencari angka NPM ({key_npm}): {'Ditemukan' if result else 'Tidak Ditemukan'}")

key_false = 98
result_false = bst.search(root_bst, key_false)
print(f"Mencari angka bebas ({key_false}): {'Ditemukan' if result_false else 'Tidak Ditemukan'}")
```
